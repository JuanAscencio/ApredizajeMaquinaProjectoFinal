***Fase 2 - Clasificación inicial con MLP***

Objetivo: Entrenar el primer modelo MLP usando TODAS las características excepto "Longitud".


1. División entrenamiento/prueba

```python
#Usaremos estratificación porque son 10 clases
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

2. Crear modelo MLP

```python
from sklearn.neural_network import MLPClassifier

mlp = MLPClassifier(
    hidden_layer_sizes=(100,),
    activation='relu',
    solver='adam',
    max_iter=500,
    random_state=42
)
```

3. Entrenar modelo

```python
mlp.fit(X_train, y_train)
```

4. Predicciones

```python
y_pred = mlp.predict(X_test)
```

5. Metricas

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print(classification_report(y_test, y_pred))
```

6. Matriz de confusión

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(10,8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title("Matriz de confusión")
plt.xlabel("Predicción")
plt.ylabel("Valor real")
plt.show()
```

7. Curva de aprendizaje

```python
plt.figure(figsize=(8,5))
plt.plot(mlp.loss_curve_)
plt.title("Curva de aprendizaje")
plt.xlabel("Iteraciones")
plt.ylabel("Loss")
plt.grid(True)
plt.show()
```
